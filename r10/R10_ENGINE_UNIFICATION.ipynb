```python
import json
from pathlib import Path
ROOT=Path.cwd().parent if Path.cwd().name=='r10' else Path.cwd()
result=json.loads((ROOT/'r10/results/r10_metrics.json').read_text())
result['data_as_of'], result['predeclared_selection']
```

## Results

```python
import pandas as pd
rows=[]
for period in ['historical_test','post_2019_stress','post_emergency']:
 for h,values in result[period]['metrics'].items():
  core=values['core']; rows.append({'period':period,'horizon':int(h),'core_rmse':core['rmse'],'persistence_rmse':values['persistence']['rmse'],'ar4_rmse':values['ar4']['rmse']})
pd.DataFrame(rows)
```

## Takeaways

The parsimonious monthly core was selected; energy and expectations augmentations were rejected. It wins at 1–6 months in the separated pre-pandemic test but loses at 12 months. It beats recorded baselines during the post-2019 stress window, while fixed pre-2020 intervals under-cover materially. These are constraints on release claims, not tuning targets.
